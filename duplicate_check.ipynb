# Minimal example of a duplicate check

```python
import numpy as np
import pandas as pd
import duckdb
pd.set_option('display.max_columns', None)
```

```python
INSURANCE_CLAIMS_PATH = 'Datasets/insurance_claims_info.csv'
INSURED_INFO_PATH = 'Datasets/insured_info.csv'

before_row_count = duckdb.sql(f'''SELECT COUNT(*) FROM read_csv('{INSURANCE_CLAIMS_PATH}');''').fetchone()[0]

result = duckdb.sql(f'''CREATE TABLE add_insured_info AS SELECT tbl1.*, tbl2.* FROM read_csv('{INSURANCE_CLAIMS_PATH}') AS tbl1 LEFT JOIN read_csv('{INSURED_INFO_PATH}') AS tbl2 ON tbl1.policy_number = tbl2.policy_number;''')

after_row_count = duckdb.sql('''SELECT COUNT(*) FROM add_insured_info;''').fetchone()[0]

if after_row_count > before_row_count:
    raise ValueError(f'Joining insured information created duplicates! Expected: {before_row_count}. Actual: {after_row_count}')
```
